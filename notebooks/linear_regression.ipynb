```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
from Regression.Regression import StandardScaler, TrainTestSplit, SimpleRegression
```

```python
california = fetch_california_housing()
```

```python
california
```

```python
california.target_names
```

```python
pd.DataFrame(california.data, columns=california.feature_names).shape
```

```python
df = pd.DataFrame(california.data, columns=california.feature_names)
df[california.target_names[0]] = california.target
df
```

```python
x = df.drop('MedHouseVal', axis=1)
x
```

```python
y = df['MedHouseVal']
y
```

```python
model = TrainTestSplit()
```

```python
x_train, y_train, x_test, y_test = model.split(x, y)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
scalar = StandardScaler()
x_train_scaled = scalar.fit(x_train)
x_train_scaled
```

```python
SR = SimpleRegression()
SR.fit(x_train_scaled, y_train)
```

```python
SR.err_sq
```

```python
x_test_scaled = scalar.transform(x_test)
x_test_scaled
```

```python
y_cap = SR.predict(x_test_scaled)
y_cap
```

```python
y_train
```

```python
y_cap
```

```python
import numpy as np
err = y_cap - y_test
err_sq = np.square(err)
```

```python
err_sq.sum()/x_test.shape[0]
```

```python
from sklearn.linear_model import LinearRegression
model_sk = LinearRegression()
```

```python
model_sk.fit(x_train_scaled, y_train)
```

```python
model_sk.score(x_train_scaled, y_train)
```

```python
y_cap_sk = model_sk.predict(x_test_scaled)
```

```python
model_sk.score(x_test_scaled, y_test)
```

```python
y_cap_sk_tr = model_sk.predict(x_train_scaled)
```

```python
err_sk_tr = np.square(y_cap_sk_tr - y_train)
mse = np.sum(err_sk_tr)/x_train_scaled.shape[0]
mse
```

```python
err_ts = np.square(y_cap_sk - y_test)
mse_ts = np.sum(err_ts)/x_test_scaled.shape[0]
mse_ts
```
