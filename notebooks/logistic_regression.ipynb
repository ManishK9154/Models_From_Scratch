```python
df_encoded = pd.get_dummies(df, columns = ['StreamingMovies', 'gender', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'Contract', 'PaymentMethod'], drop_first=True)
df_encoded = df_encoded.replace('No',0).replace('Yes',1)
df_encoded = df_encoded.replace(True, 1).replace(False, 0)
df_encoded
```

```python
df_encoded.drop(['customerID'], axis=1, inplace=True)
```

```python
df_encoded
```

```python
x = df_encoded.drop('Churn', axis=1)
y = pd.get_dummies(df_encoded['Churn'], drop_first=True)
```

```python
x.loc[x['TotalCharges']==' ', 'TotalCharges'] = '0'
x[x['TotalCharges']==' ']
```

```python
x['TotalCharges'] = x['TotalCharges'].astype(float)
x['TotalCharges']
```

```python
for c in x.columns:
    if x[c].dtype=='object':
        x[c] = x[c].astype(int)
```

```python
y = pd.Series([1 if x==True else 0 for x in y[1]])
y
```
